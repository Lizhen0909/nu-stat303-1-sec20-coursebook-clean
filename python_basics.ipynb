## Variables, types, and expressions

A variable name refers to an object. Assignment binds a name to an object; later assignment can bind the same name to a different object, even one of a different type.

Use descriptive `snake_case` names such as `temperature_c` and `valid_count`. Names are case-sensitive, cannot start with a digit, and cannot be reserved keywords such as `for`. Letters, digits, and underscores cover the names we will use; Python also permits many Unicode letters. Avoid hiding built-ins with names such as `list`, `str`, or `sum`.


````python
temperature_c = 18.5
station_count = 3
is_valid = True
station_name = "North"
missing_reading = None

for value in [temperature_c, station_count, is_valid, station_name, missing_reading]:
    print(repr(value), type(value).__name__)
````

Here, `int`, `float`, `bool`, and `str` are types; `None` is a special value whose type is `NoneType`. It commonly represents the absence of a value. A numeric zero can be a valid measurement and should not automatically be treated as missing.

Lists, tuples, sets, and dictionaries organize collections of values. Strings are immutable text sequences. We will review these structures in the next chapter.

### Convert deliberately

Text that looks numeric is still text. `float()` and `int()` convert suitable values; `str()` creates a text representation. Conversion can fail when the content does not match the requested type.


````python
reading_text = "18.5"
reading_value = float(reading_text)
print(reading_value + 2)
print(reading_text + "2")
````

The first operation adds numbers; the second joins strings. `float("unknown")` would raise `ValueError`. Likewise, `int("18.5")` fails: that text is not an integer literal. Decide what a value means before converting it.

## Conditions and decisions

A condition determines which code runs. Indent the body consistently with **four spaces**. Configure your editor to insert spaces when you press Tab; mixing tabs and spaces can cause errors.


````python
reading = 105

if reading is None:
    status = "missing"
elif 0 <= reading <= 100:
    status = "within the expected range"
else:
    status = "needs investigation"

print(status)
````

Python checks branches in order and runs the first matching branch. The `else` branch runs if none match. The range here is an example rule, not a universal definition of valid data: a real analysis needs a rule appropriate to the measurement.

Use `is None` to check for the special missing-value placeholder. Use `==` to compare ordinary values. Checking for `None` first avoids trying to compare it with a number.

### Combine conditions

- `and`: both conditions must hold.
- `or`: at least one condition must hold.
- `not`: reverses a truth value.

Python short-circuits `and` and `or`: it evaluates the second expression only when needed.


````python
reading = None
is_acceptable = reading is not None and 0 <= reading <= 100
print(is_acceptable)

reading = 105
outside_range = reading < 0 or reading > 100
print(outside_range)
print(not outside_range)
````

### Truth values are not data-quality rules

Common false values include `False`, `None`, numeric zeros, and empty strings and collections. Other objects can define their own truth behavior; this is not an exhaustive list. `bool()` is a built-in function for checking truth values.


````python
print(bool(0))
print(bool(None))
print(bool(""))
print(bool("0"))
````

`"0"` is nonempty text, so it is true. For numerical measurements, `if reading:` cannot distinguish a valid zero from `None`. Use an explicit check that matches your question. Later, arrays and tables will require their own tools for combining conditions and checking missing data.

## Loops and repeated work

### Start with a `for` loop

A `for` loop visits each item in an **iterable**, such as a list, string, or range. You do not need a counter just to access each item.


````python
readings = [18.5, None, 21.0, -2.0, 0.0]
valid_count = 0
valid_total = 0.0

for reading in readings:
    if reading is not None and 0 <= reading <= 100:
        valid_count += 1
        valid_total += reading

print(f"Valid readings: {valid_count}")
print(f"Total: {valid_total}")
if valid_count > 0:
    print(f"Mean: {valid_total / valid_count:.2f}")
````

The loop includes the valid zero. It excludes missing and out-of-range readings using an explicit rule. Checking the count before division handles the possibility of no valid readings.

**Modify:** Replace `readings` with `[None, -2.0]`. What should the program report? What happens if the guard before division is removed?

::: {.callout-note collapse="true" title="Expected result"}
The count is `0` and the total is `0.0`; no mean is printed. Without the guard, division by zero raises `ZeroDivisionError`. For a report, you could add an `else` branch displaying `"No valid readings"`.
:::

### Use `range()` when you need integer steps

The stop value is always **excluded**. A positive step moves upward; a negative step moves downward. If the direction cannot reach any value before the stopping boundary, the range is empty. A step of zero is invalid.


````python
print(list(range(4)))
print(list(range(1, 4)))
print(list(range(1, 8, 2)))
print(list(range(3, 0, -1)))
````

For an item and its position, `enumerate()` is usually clearer than indexing through `range(len(...))`. Data Structures reviews that pattern in more detail.

### `break` and `continue`

`break` exits the nearest enclosing loop. `continue` skips the remaining statements in the current iteration and starts the next iteration.


````python
# Stop at the first reading that needs investigation.
for reading in [18.5, 105.0, 21.0]:
    if reading > 100:
        print(f"Investigate: {reading}")
        break
    print(f"Checked: {reading}")
````

````python
# Skip missing entries while keeping valid zeros.
for reading in [18.5, None, 0.0]:
    if reading is None:
        continue
    print(f"Recorded: {reading}")
````

**Predict:** Why is `21.0` not checked in the first example? Why is `0.0` printed in the second?

::: {.callout-note collapse="true" title="Explanation"}
`break` ends the first loop when `105.0` is encountered. In the second loop, only `None` triggers `continue`; zero remains a valid entry.
:::

### Use a `while` loop for a stopping condition

A `while` loop repeats while its condition is true. Identify what changes on each iteration and why the loop will eventually stop.


````python
# Compute 10! = 1 * 2 * ... * 10.
n = 10
factorial = 1
factor = 1

while factor <= n:
    factorial *= factor
    factor += 1

print(f"{n}! = {factorial}")
````

## Functions that return results

A function gives a reusable operation a name. It can accept zero or more arguments. Defining a function does not run its body; calling it does.

### `return` versus `print`

`print()` displays information. `return` sends a value back to the caller so another calculation can use it. A function that reaches its end without returning a value returns `None`.


````python
def show_fahrenheit(celsius):
    print(celsius * 9 / 5 + 32)

shown_result = show_fahrenheit(20)
print(f"Stored result: {shown_result}")
````

````python
def celsius_to_fahrenheit(celsius):
    """Return a Celsius temperature converted to Fahrenheit."""
    return celsius * 9 / 5 + 32

converted = celsius_to_fahrenheit(20)
print(f"Converted: {converted}")
print(f"Five degrees higher: {converted + 5}")
````

### Parameters, keyword arguments, and defaults

A **parameter** is a name in a function definition. An **argument** is a value supplied in a call. Arguments may be supplied by position or by keyword. A parameter with a default may be omitted from the call.


````python
def is_valid_reading(value, lower=0, upper=100):
    """Return whether a numeric reading is present and within the limits."""
    return value is not None and lower <= value <= upper

print(is_valid_reading(18.5))
print(is_valid_reading(None))
print(is_valid_reading(105, upper=120))
print(is_valid_reading(value=-2, lower=-10, upper=40))
````

This function expects a number or `None`; it does not convert arbitrary text. Documenting that expectation helps the caller use it correctly. Keyword arguments make choices such as `upper=120` visible at the call site.

### Local names and global names

Parameters and names assigned inside a function are normally local to that function. A function can read names from its module's global scope, but passing required values as arguments makes dependencies easier to see.


````python
station_label = "North"  # Global name in this notebook.

def display_station():
    print(station_label)  # Reads the global name; does not declare a local one.

def make_station_label(name):
    cleaned_label = name.strip().title()  # Local name.
    return cleaned_label

display_station()
print(make_station_label("  south  "))
````

`cleaned_label` is not made available outside `make_station_label`; its value is made available through `return`. Assignment to a name inside a function normally creates or updates a local binding rather than rebinding a global name. Shared mutable objects are a separate issue, reviewed in [Data Structures](data_structures.ipynb#shared-references-and-mutation).

::: {.callout-note collapse="true" title="Optional review: flexible argument lists"}
In documentation, `*args` collects extra positional arguments into a tuple and `**kwargs` collects extra keyword arguments into a dictionary. The `*` and `**` syntax gives these names their roles; the names themselves are conventions.

```python

    print(args)
    print(kwargs)

describe_call(18.5, 21.0, unit="C")
# (18.5, 21.0)
# {'unit': 'C'}
```

For this review, prioritize ordinary parameters, returned values, and keyword arguments. You do not need to design flexible interfaces before starting STAT 303-1.

## Objects, attributes, and methods

Names refer to objects. An object's **attributes** are accessed using a dot; its **methods** are operations associated with the object. A method is itself an attribute that you can call.


````python
observed_at = dt.datetime(2025, 9, 20, 11, 30)
print(observed_at.year)                    # Read an attribute.
print(observed_at.strftime("%Y-%m-%d"))    # Call a method.

station_label = "north"
print(station_label.upper())
print(station_label)                      # Strings are immutable.
````

Parentheses call the method. `station_label.upper()` returns a new string; it does not change the existing string. Later, library objects will also expose attributes and methods. Learn to check what an operation returns and whether it changes its object.
