## What Is a Terminal?

A **terminal** is a text interface where you type commands and read their output. Instead of clicking a folder or a button, you describe an action in text. A **shell** is the program that reads those commands and carries them out. Common shells include zsh or Bash on macOS, and PowerShell or Command Prompt on Windows.

If you typed a command such as `quarto render assignment.ipynb --to html` in STAT 201, you have already used a terminal: you asked the shell to start Quarto, gave it a notebook filename, and requested HTML output. You do not need to run this example now.

In VS Code, open **Terminal → New Terminal**. This integrated terminal lets you work beside your notebook without opening a separate application. Check its tab or dropdown for the shell name; this chapter provides separate commands where shells differ. See the [VS Code terminal introduction](https://code.visualstudio.com/docs/terminal/getting-started) for more about the interface.

### Read the Prompt and Enter a Command

The **prompt** is the text that appears when the shell is ready for input. Yours might look like one of these examples:

```text
student@laptop stat303-setup %
PS C:\Users\student\stat303-setup>
```

Type after the prompt and press **Enter** to run a command. Do not copy the prompt itself. Read any output, then wait for the prompt to return before entering the next command. Some successful commands produce no output.

A terminal can also run an interactive Python session. If you see `>>>`, Python is waiting for Python code; type `exit()` and press Enter to return to the shell before using the navigation commands below. In a notebook, Python code goes in a **code cell** and runs through the selected **kernel**.

## Why Use the Terminal for Python Data Science?

Your notebook holds calculations, plots, and explanations. The terminal helps you prepare and run the project around it:

- **Manage Python environments and packages.** Create a separate environment for the project and install libraries such as Pandas into it.
- **Locate files.** Check which folder you are working in and whether it contains the notebook or dataset a command needs.
- **Run project tools.** Render an HTML report with Quarto or run a saved Python script.
- **Repeat a workflow.** Save the commands you used so that you or a classmate can repeat the setup later.

For example, a data analysis assignment may involve installing Pandas in the terminal, analyzing a CSV in a notebook, and returning to the terminal to render the saved notebook as HTML. Keeping these steps in one VS Code workspace makes the workflow easier to follow.


## Navigate in the Terminal

The terminal always has a **current working directory**, meaning the folder from which it interprets relative file and folder names. Here, *directory* and *folder* mean the same thing. Think of it as your current location in the folder tree: listing files shows what is here, and `cd` (change directory) moves you somewhere else.

With the extracted `stat303-setup` folder open in VS Code, open a terminal and use the column for your shell. Windows PowerShell and Windows Command Prompt are different shells. Run one command at a time.

| Task | macOS zsh/Bash | Windows PowerShell | Windows Command Prompt |
|---|---|---|---|
| Show current directory | `pwd` | `Get-Location` | `cd` |
| List files and folders | `ls` | `Get-ChildItem` | `dir` |
| Enter a child folder | `cd data` | `cd data` | `cd data` |
| Move to parent folder | `cd ..` | `cd ..` | `cd ..` |

`.` means the current directory and `..` its parent. A **child folder** is a folder directly inside the current one; the **parent folder** is the folder containing it. Put paths containing spaces in quotes, for example `cd "course practice"` when that folder exists inside your current directory. To change drives as well as folders in Command Prompt, use `cd /d "D:\your\folder"`, replacing the example path with your actual path.

## Ask Three Questions Before a Command

Now that you can find your way around, build this habit before running setup or analysis commands. The answers determine how a command is interpreted, which Python installation it uses, and where it looks for or writes files.

| Question | In a notebook | In a terminal |
|---|---|---|
| **1. Where am I typing?** | Use a Python code cell for Python code. | Identify the shell; a `>>>` prompt means you are inside Python instead. |
| **2. Which Python am I using?** | Run `import sys`, then `print(sys.executable)`. | Run `python -c "import sys; print(sys.executable)"`. |
| **3. Which folder is current?** | Run `from pathlib import Path`, then `print(Path.cwd())`. | Use `pwd`, `Get-Location`, or `cd` for your shell. |

The terminal command `python -c "import sys; print(sys.executable)"` starts Python, runs the code inside the quotes, and exits back to the shell. On macOS, your installation may use `python3`; use that in place of `python` when needed. If neither command works, return to the previous chapter's Python setup checks. Navigation commands such as `cd` do not use Python, so question 2 is **not applicable** for those commands.

The notebook and terminal can use **different Python environments and different working directories**. The folder visible in Explorer does not prove either process's current directory. An active-environment prompt is a hint, not proof that the notebook kernel matches.

## Why Use a Python Virtual Environment?

In Chapter 1, you selected Anaconda's **`base` environment** from STAT 201 to check that your notebook could run in VS Code. That gave you a familiar starting point. As you begin new projects, use a **separate environment for each project** instead of installing everything into `base`.

### Different Projects Need Different Tools

A computer can have **multiple Python versions** installed. One project might require Python 3.11, while another uses Python 3.12. These are examples, not instructions to install both versions for this course. Code and packages written for one version may need changes to work with another.

Projects also use different **libraries**, installed as packages. For example, Pandas helps analyze tables, while Matplotlib helps create plots. Each package has its own versions, separate from Python's version, and may depend on other packages. These required packages are called **dependencies**.

Conflicts can arise when two tools need incompatible versions of the same dependency. As a hypothetical example, an older analysis tool might require version 1 of a shared library, while a newer tool requires version 2. Installing or upgrading packages for one project in a shared environment can make another project's previously working code fail.

### Give Each Project Its Own Environment

A **Python virtual environment** gives a project access to a particular Python interpreter and its own set of installed packages. Think of it as a toolbox assigned to that project. Two projects can use different Python versions, different packages, or different versions of the same package. They can also use the same Python version while keeping their packages separate.

![Each project uses its own Python environment. The versions and shared library shown are hypothetical examples, not course installation requirements.](images/python-project-environments.svg){#fig-python-project-environments width=100% fig-alt="Two projects on one computer use separate environments. Project A uses Python 3.11 and shared library version 1; Project B uses Python 3.12 and shared library version 2. Each notebook connects to its own environment, so changing packages in one does not change the other. Anaconda base remains the familiar STAT201 setup."}

With separate environments, changing packages in Project B does not change the packages installed in Project A. Isolation does not make incompatible libraries work together inside one environment; each project still needs a compatible set of dependencies.

**For STAT303:** keep the notebooks in `stat303-setup` together as one course project and give that project one environment named `.venv`. You do not need a new environment for every notebook. Later, an unrelated project should get its own environment. In VS Code, connect each notebook to its project's environment through the kernel picker.

::: {.callout-note title="Python version, environment, and kernel"}
The **Python version** identifies the language/interpreter release. The **environment** provides that interpreter and a set of packages. The **kernel** is the running process that executes your notebook's code using the selected environment.

Python's built-in `venv` tool creates an environment from an **already installed interpreter**. It does not download a different Python version, and by default it does not include packages installed in `base`. You will install the course packages into the new environment. See the [Python virtual environment tutorial](https://docs.python.org/3/tutorial/venv.html).
:::

Now that you know why projects need separate environments, create one for this course and verify that your terminal and notebook both use it.

## Install Packages in the Intended Environment

### From the Terminal

**Starting folder: `stat303-setup`; verified `.venv` Python active.**

```text
python -m pip install ipykernel numpy pandas matplotlib
```

If using the PowerShell direct-executable route, replace `python` with `.\.venv\Scripts\python.exe` in these commands. `python -m pip` uses the pip belonging to that Python; bare `pip` can be ambiguous.

### From a Notebook

**Run in a code cell after selecting `.venv`:**

```python
%pip install seaborn
```

`%pip` targets the current IPython kernel's environment. Prefer the explicit `%` over relying on automagic. `!pip` invokes a shell command and may find a different pip. Restart the notebook kernel after installation when needed, then verify:

```python
import sys
import pandas as pd
import seaborn as sns
print(sys.executable)
print(pd.__version__)
print(sns.__version__)
```

Compare the notebook executable with the terminal's executable. A successful terminal installation does not prove that another notebook kernel can import the package. Keep installation work in a temporary setup cell or terminal, rather than rerunning installations every time an assignment is executed.


## Find Local Data with Paths

An **absolute path** begins at the filesystem root (including the appropriate drive on Windows). A **relative path** is interpreted from the process's current working directory. Forward slashes alone do not make an absolute path portable to another person's computer.

Use `pathlib` to build paths without manually joining platform-specific separators. In a **notebook code cell**:

```python
from pathlib import Path
print(Path.cwd())
data_path = Path("data") / "station_readings.csv"
print(data_path.resolve())
print(data_path.is_file())
```

For these root-level practice notebooks, the expected working directory is the extracted `stat303-setup` folder. If `is_file()` is false, inspect the displayed directory, filename, capitalization, and extracted folder layout. VS Code settings or a different notebook location can change the working directory; always check rather than assume.

Once it is true, inspect a few lines without needing data-analysis syntax:

```python
print(data_path.read_text(encoding="utf-8").splitlines()[:3])
```

A path expression describes a location; it does not establish that a file exists. `.exists()` checks for an existing path, while `.is_file()` checks specifically for a file.

### What Changes When a Notebook Moves?

If a notebook runs with `stat303-setup/notebooks` as its working directory, `Path("data")` refers to a folder inside `notebooks`. The sibling data folder would instead be:

```python
Path("..") / "data" / "station_readings.csv"
```

This works only when the actual working directory has that relationship. Do not repeatedly add `..` until a path happens to work.

The terminal's `cd` changes that terminal's directory, not an already running notebook's directory. A notebook's `%cd` changes its working directory persistently; `!cd` runs in a temporary shell and generally does not persist. For the main workflow, keep a consistent layout and explicit paths rather than scattering directory changes through the notebook.


## Record and Recreate Dependencies

**Terminal, starting in `stat303-setup`, with the project Python verified:**

```text
python -m pip freeze > requirements.txt
```

The shell writes `requirements.txt` into its current directory. `>` replaces an existing file with that name. Review the file and confirm that it records your project environment rather than Anaconda's entire base environment.

In a new compatible environment, install from it with:

```text
python -m pip install -r requirements.txt
```

Package versions help reproduce an environment but do not guarantee identical behavior across Python versions, operating systems, or external services. Record the Python version too. Incompatible requirements can cause pip's resolver to report an error; it does not simply select the last version mentioned.
