```python
from pathlib import Path
data_path = Path("data") / "station_readings.csv"
print("Resolved data path:", data_path.resolve())
print("File found:", data_path.is_file())
if not data_path.is_file():
    raise FileNotFoundError("Check the working directory and extracted data folder before continuing.")
```

```python
import pandas as pd
import matplotlib
print("Pandas:", pd.__version__)
print("Matplotlib:", matplotlib.__version__)
```

```python
data = pd.read_csv(data_path)
display(data)
print("Records:", len(data))
print("Mean temperature (C):", data["temperature_c"].mean())
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.bar(data["station_id"], data["temperature_c"])
ax.set_title("Illustrative Station Temperatures")
ax.set_xlabel("Station ID")
ax.set_ylabel("Temperature (degrees C)")
fig.tight_layout()
plt.show()
```
