## Choosing a Data Structure

Choose a structure according to the operations and information you need, rather than memorizing syntax alone.

| Structure | Order | Mutable? | Duplicates | Access | Typical use |
|---|---|---|---|---|---|
| `list` | Positional order | Yes | Allowed | Index or slice | A sequence of readings or records |
| `tuple` | Positional order | No | Allowed | Index or slice | A record with a fixed arrangement of fields |
| `str` | Character order | No | Allowed | Index or slice | Text |
| `set` | No guaranteed order | Yes | Removed | Membership; no indexing | Unique identifiers and overlap checks |
| `dict` | Insertion order | Yes | Keys unique; values may repeat | Key | Labeled fields or identifier-based lookup |

A tuple's immutability prevents replacing its elements; it does not make a nested mutable object immutable. Dictionary keys and set elements must be **hashable**. Strings and integers are common choices. A tuple is hashable only when all its elements are hashable: `(101, "STAT303")` works, but `([101], "STAT303")` does not.


````python
readings = [20, 20, 18]
station_location = (42.05, -87.68)
station_name = "North"
unique_readings = set(readings)
station = {"id": "N01", "name": station_name}
print(readings)
print(sorted(unique_readings))  # Sort for a predictable display, not arrival order.
print(station_location)
print(station)
````

::: {.callout-note collapse="true" title="Suggested choices"}
Use a list for the history, a set for unique IDs, and a dictionary for named fields. Converting the history to a set would discard repeated measurements and their original order.

## Accessing and Updating Values

### Indexing and Slicing

Sequence indexes start at zero; negative indexes count from the end. A slice excludes its stop position. Out-of-range indexing raises `IndexError`, whereas a slice can safely extend beyond the end.


````python
values = [10, 20, 30, 40]
print(values[0])
print(values[-1])
print(values[1:3])
print(values[:2])
print(values[::2])
print(values[1:100])
````

### Updating Lists

`append()` adds **one object**. `extend()` adds each item from an iterable. These methods change the list in place and return `None`.


````python
appended = [10, 20]
append_result = appended.append([30, 40])
extended = [10, 20]
extended.extend([30, 40])
print(appended)
print(extended)
print(append_result)
extended[0] = 11
print(extended)
````

**Predict:** What would `extended.extend("50")` add? Why is that different from `extended.append("50")`?

::: {.callout-note collapse="true" title="Answer"}
`extend("50")` adds two separate strings, `"5"` and `"0"`, because a string is iterable. `append("50")` adds one string, `"50"`. Neither operation converts text into a number.
:::

### Working with Text

String methods return new strings rather than changing the original string. `strip()` removes leading and trailing whitespace; `split()` separates text into pieces; a separator's `join()` method joins an iterable of strings.


````python
raw_label = "  north, temperature  "
parts = raw_label.strip().split(",")
clean_parts = []
for part in parts:
    clean_parts.append(part.strip())
print(raw_label)
print(clean_parts)
print(" | ".join(clean_parts))
````

`split(",")` is suitable for this simple label. CSV files can contain quoted commas and other rules, so use the standard-library `csv` module or a data-reading library for actual CSV data.

### Dictionary Keys and Missing Values

`record[key]` retrieves a value or raises `KeyError` if the key is absent. `record.get(key, default)` returns the default for an absent key; without a supplied default it returns `None`. Membership in a dictionary tests **keys**, not values.


````python
station = {"id": "N01", "reading": None}
station["unit"] = "C"
station["unit"] = "degrees C"  # Update the existing key.
print(station["id"])
print("reading" in station)
print("N01" in station)
print(station.get("reading", "not recorded"))
print(station.get("height", "not recorded"))
````

A present key whose value is `None` is different from an absent key. `.get()` does not replace an existing `None` with its default.

### Sets and Membership

Use sets to find unique values and compare groups. Union finds values in either set; intersection finds shared values; difference finds values in one set but not the other.


````python
expected_ids = {"N01", "S01", "E01"}
received_ids = {"N01", "S01", "W01"}
print("All IDs:", sorted(expected_ids | received_ids))
print("Shared IDs:", sorted(expected_ids & received_ids))
print("Missing IDs:", sorted(expected_ids - received_ids))
print("Unexpected IDs:", sorted(received_ids - expected_ids))
````

Dictionary key lookup and set membership are typically fast on average; list membership searches through values. Do not confuse membership with positional indexing: `values[0]` is direct list access, while sets do not support indexing. For this course, choose the correct representation before considering performance.

## Shared References and Copying {#shared-references-and-mutation}

Assignment binds a name to an object; it does not automatically copy that object. **Mutation** changes an object. **Rebinding** changes which object a name refers to.


````python
original_readings = [18.5, 21.0]
shared_readings = original_readings
shared_readings.append(0.0)
print(original_readings)
print(shared_readings is original_readings)
shared_readings = [99.0]
print(original_readings)
print(shared_readings)
````

### Shallow Copies

`list.copy()` makes a new outer list. Appending to this copy does not change the original outer list, but objects nested inside both lists remain shared.


````python
independent_readings = original_readings.copy()
independent_readings.append(30.0)
print(original_readings)
print(independent_readings)

nested = [[1], [2]]
shallow_copy = nested.copy()
shallow_copy[0].append(3)
print(nested)
````

For nested structures, the standard-library `copy.deepcopy()` can recursively copy objects when independent nested data is required. Do not copy automatically: decide whether sharing is intended.

### Function Arguments

A parameter refers to the object supplied by the caller. Mutating a passed list is visible to the caller; rebinding the parameter does not rebind the caller's name. This behavior is often called *call by sharing*.


````python
def append_zero(values):
    values.append(0.0)

def replace_locally(values):
    values = [99.0]
    return values

measurements = [18.5]
append_zero(measurements)
print(measurements)
print(replace_locally(measurements))
print(measurements)
````

**Explain:** When could an analysis be affected by appending to a list inside a function? How could you return an expanded list while preserving the caller's list?

::: {.callout-note collapse="true" title="One solution"}
Later calculations may unexpectedly include the appended value. For a flat list, create a new list:

```python
def with_zero(values):
    result = values.copy()
    result.append(0.0)
    return result

source = [18.5]
expanded = with_zero(source)
print(source)    # [18.5]
print(expanded)  # [18.5, 0.0]
```
:::

## Iteration and Unpacking

### Collections and Iterators

An **iterable** can supply an iterator to a loop. An **iterator is also iterable**: it remembers its position and becomes exhausted after producing its values. A list can supply fresh iterators for repeated passes; an already exhausted iterator does not restart itself.


````python
readings = [18.5, 21.0]
reading_iterator = iter(readings)
print(next(reading_iterator))
print(list(reading_iterator))
print(list(reading_iterator))
print(list(readings))
print(iter(reading_iterator) is reading_iterator)
````

Calling `next()` on an exhausted iterator raises `StopIteration`; a `for` loop handles that signal automatically. `next(iterator, default)` can provide a fallback instead.

### Unpacking and `enumerate()`

Unpacking assigns items to names. Ordinary unpacking needs the same number of names and items; starred unpacking collects remaining items into a list. `enumerate()` supplies a counter and an item. Its `start` argument changes the counter, not the underlying sequence indexes.


````python
station_id, reading = ("N01", 18.5)
print(station_id, reading)
first, *rest = [10, 20, 30]
print(first, rest)
for row_number, label in enumerate(["North", "South"], start=1):
    print(row_number, label)
````

For dictionaries, use `for key, value in record.items()` when you need both labels and values.

### Pairing with `zip()`

`zip()` returns an iterator of tuples paired by **position**. It does not match identifiers or verify that two inputs refer to the same entities. By default, it stops at the shortest input.


````python
station_ids = ["N01", "S01", "E01"]
readings = [18.5, 21.0]
pairs = list(zip(station_ids, readings))
print(pairs)
print("Lengths match:", len(station_ids) == len(readings))

paired_ids, paired_readings = zip(*pairs)
print(paired_ids)             # A tuple, not a list.
print(list(paired_readings))  # Convert explicitly if a list is needed.
````

Here `E01` has no partner and is omitted. If equal-length lists are required, check their lengths before pairing. Python 3.10 and later also support `zip(..., strict=True)`, which raises `ValueError` for mismatched lengths when consumed. Neither approach detects IDs in different orders. The unzip pattern above also assumes there are pairs to unpack.

### Useful Functions and Their Limits

| Tool | Purpose | Important detail |
|---|---|---|
| `len()` | Count items in a sized object | An arbitrary iterator may have no length |
| `min()`, `max()` | Find extrema | Empty input raises `ValueError` unless an appropriate `default` is supplied |
| `sum()` | Total numerical values | An empty input returns `0` with the default starting value |
| `sorted()` | Create a sorted list | Consumes an iterator supplied as input |
| `reversed()` | Iterate in reverse | Requires a reversible object; not every iterable supports it |
| `all()`, `any()` | Combine truth tests | `all([])` is `True`; `any([])` is `False` |
| `separator.join()` | Join text items | All items must be strings |

## Comprehensions

Use a comprehension for a simple transformation or filter. Start with an ordinary loop, then compare the equivalent expression. Use a loop when several steps or branches make the comprehension difficult to read.


````python
readings = [18.5, None, 21.0, -2.0, 0.0]
valid_readings = []
for value in readings:
    if value is not None and 0 <= value <= 100:
        valid_readings.append(value)

valid_comprehension = [
    value for value in readings
    if value is not None and 0 <= value <= 100
]
print(valid_readings)
print(valid_comprehension)
````

The expression comes first, then the iteration, then an optional filter. Square brackets build a list, braces around a single expression build a set, and braces around `key: value` build a dictionary.


````python
squares = [number ** 2 for number in range(5, 9)]
labels = ["north", "south", "north"]
unique_labels = {label.upper() for label in labels}
label_lengths = {label: len(label) for label in labels}
print(squares)
print(sorted(unique_labels))
print(label_lengths)
````

The set loses repetitions. The dictionary has only one value per key; repeated keys replace earlier values. Neither structure can preserve repeated records under the same key without additional design.

## Sorting and Key Functions

### New List or In-Place Change?

`sorted()` returns a new list. A list's `.sort()` changes that list and returns `None`. If you pass an iterator to `sorted()`, the iterator is consumed even though a new list is returned.


````python
values = [3, 1, 2]
ordered = sorted(values)
print(values)
print(ordered)
result = values.sort()
print(values)
print(result)
print(sorted(values, reverse=True))
````

### Choosing a Key

The `key` argument accepts a function that extracts each item's sorting value. Pass the function itself; do not call it at the argument position. Equal keys retain their original relative order: Python sorting is **stable**.


````python
words = ["pear", "apple", "banana", "kiwi"]
print(sorted(words, key=len))

def last_character(word):
    return word[-1]

print(sorted(words, key=last_character))
print(sorted(words, key=lambda word: word[-1]))
````

`pear` stays before `kiwi` when sorting by length because both have four letters and `pear` came first. The last-character example assumes nonempty strings.

A **lambda** creates a function from a single expression. It returns that expression's value. Use it for a short key such as `lambda record: record["reading"]`; use `def` when a meaningful name or multiple steps improve clarity.


````python
number_text = ["10", "2", "1"]
print(sorted(number_text))
print(sorted(number_text, key=int))

people = [("Chen", 22), ("Bob", 19), ("Ann", 22)]
print(sorted(people, key=lambda person: person[1]))
print(sorted(people, key=lambda person: (person[1], person[0])))
````

The first people sort preserves Chen-before-Ann for equal ages. The second explicitly breaks age ties alphabetically by name. Say which behavior you intend in a task.

::: {.callout-note collapse="true" title="Optional: map, filter, and reduce"}
`map()` transforms items and `filter()` selects items; both return iterators. A simple comprehension often expresses the same task more directly. `functools.reduce()` repeatedly combines values; prefer a clear built-in operation when one exists.

```python
from functools import reduce
values = [1, 2, 3, 4]
print(list(map(lambda value: value ** 2, values)))
print([value ** 2 for value in values])
print(list(filter(lambda value: value % 2 == 0, values)))
print([value for value in values if value % 2 == 0])
print(reduce(lambda left, right: left * right, values))  # 24
```
Recognizing these tools is useful; mastering them is not a prerequisite for starting the data science libraries.

## Working with Structured Records

A **list of dictionaries** keeps each observation's fields together and gives the fields names. Nested access combines sequence indexing with dictionary lookup.


````python
records = [
    {"station": "N01", "reading": 18.5},
    {"station": "S01", "reading": 21.0},
    {"station": "N01", "reading": 19.0},
]
print(records[0]["reading"])
print([record["reading"] for record in records if record["station"] == "N01"])
print(sorted(records, key=lambda record: record["reading"]))

latest_by_station = {record["station"]: record["reading"] for record in records}
print(latest_by_station)
````

The dictionary retains only the last encountered reading per station, not every observation. Calling it "latest" is justified only if records are in time order. Here we know only their list order.

### Preserve First-Occurrence Order

A set can track names already seen while a list preserves the desired output order.


````python
names = ["Alice", "Bob", "Alice", "Charlie", "Bob", "Dave"]
seen = set()
unique_names = []
first_positions = []
for index, name in enumerate(names):
    if name not in seen:
        seen.add(name)
        unique_names.append(name)
        first_positions.append((index, len(name)))
print(unique_names)
print(sorted(first_positions, key=lambda pair: pair[1], reverse=True))
````

The positions refer to the original list, not positions after deduplication. For equal lengths, stable sorting preserves encounter order. In real data, identical names do not necessarily identify the same person; use a suitable identifier when available.
