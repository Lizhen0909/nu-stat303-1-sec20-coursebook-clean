## Anaconda vs. VS Code

Anaconda and Visual Studio Code serve different roles: Anaconda provides Python, packages, and environment management tools, while VS Code offers a flexible workspace for writing code and managing projects. The table below compares their purposes and key features.

::: {.tools-comparison}

|---|---|---|
| **Overview** | A Python distribution with environment-management tools | A powerful, flexible, and lightweight code editor |
| **What it is** | A distribution that includes Python, many common packages, and tools such as conda and Navigator | A free code editor from Microsoft, built on the open-source Code - OSS project |
| **Main purpose** | Makes it easy to install Python and manage packages and environments | Provides a modern, flexible environment to write, run, and debug code in Python and many other languages |
| **Key features** | Preinstalls many popular data science packages<br><br>Environment management with conda<br><br>Anaconda Navigator graphical interface<br><br>Works on Windows, macOS, and Linux | Clean, customizable interface<br><br>Python support through extensions<br><br>Integrated terminal and Jupyter notebook support through the Jupyter extension<br><br>Debugging, code navigation, and Git tools<br><br>Extensible through additional extensions<br><br>Works with different Python environments, including Anaconda environments |

:::

## Code, Markdown, and Saved Outputs

Use code cells for computation and Markdown cells for explanations, headings, and answers. Run cells with their play controls or `Shift+Enter`. Save the notebook with `Ctrl+S` or `Cmd+S`.

A notebook remembers values from previously executed cells. Moving or editing a cell does not undo its earlier effects. Before considering work complete, **restart the kernel, run all code cells in order, and save again**. A successfully executed notebook should not depend on variables created in deleted cells.

The HTML submitted later will be generated from this saved notebook. Writing an explanation in a code comment is useful, but Markdown is better for the report's narrative.


## If Something Does Not Work

| Symptom | First check |
|---|---|
| No kernel is offered | Python/Jupyter extensions are enabled; a Python interpreter is installed and discoverable |
| Code will not run | The cell is Code rather than Markdown and a kernel is selected |
| Output differs after restarting | All required assignments/imports occur earlier in the notebook |
| You cannot find the data folder | The ZIP is extracted and the correct project folder is open |
