## Anaconda vs. VS Code

In STAT 201, you used Anaconda to work with Jupyter notebooks. In STAT 303-1, your projects will combine notebooks, data files, installed libraries, and HTML reports. **VS Code brings these parts of a project into one workspace**, making it a good fit for the workflow you will learn in this course.

Anaconda and VS Code serve different roles:

| Tool | What it provides | How it fits your work |
|---|---|---|
| Anaconda | A Python distribution with packages and environment-management tools; Navigator provides a graphical interface for managing environments and launching applications | Supplies the Python environment you may already have from STAT 201 |
| VS Code | An editor that supports notebooks through extensions and includes a file explorer and terminal | Provides the workspace where you organize files, write code, and issue project commands |

See [Anaconda Navigator](https://www.anaconda.com/products/navigator) for its application and environment-management features.

**Why use VS Code in STAT 303-1?**

- **Keep the whole project visible.** Open a folder and work with notebooks, Python scripts, data, and report settings together. This helps you see where files belong as an assignment grows.
- **Continue using familiar notebooks.** Open your existing `.ipynb` files, write Markdown, run cells, and view outputs in the editor.
- **Get help while coding.** With the Python and Jupyter extensions, code completion, function information, and debugging tools help you explore unfamiliar libraries and investigate errors.
- **Use a terminal beside your notebook.** Install packages and run Quarto commands without switching to a separate terminal application. You still need to verify its current folder and Python environment.
- **Build a repeatable submission workflow.** Edit and run your notebook, save it, then use the integrated terminal to render your HTML with Quarto. The same workspace supports each stage of the assignment.

These features are described in the official [VS Code notebook guide](https://code.visualstudio.com/docs/datascience/jupyter-notebooks) and [integrated terminal guide](https://code.visualstudio.com/docs/terminal/basics).

**You can use VS Code with your existing Anaconda Python.** Changing editors does not require changing your notebook format or uninstalling Anaconda. In this chapter, you will select a familiar Python kernel; in Chapter 2, you will create a separate project environment. VS Code's advantage here is the integrated workflow—it does not automatically make Python code run faster or keep environments and paths correct for you.


## Know Your Tools

| Tool | Its role in this course |
|---|---|
| Python | Executes your Python code |
| Anaconda | A distribution that may already provide your Python and packages |
| Jupyter notebook | A document containing code, Markdown, and saved outputs |
| Kernel | The running process that executes a notebook's code |
| VS Code | The editor in which you open folders and work on notebooks |
| Terminal | A place to issue commands to a shell such as PowerShell, zsh, or Bash |
| Quarto | Converts your saved notebook into a formatted HTML document |

Installing VS Code does not replace Python, and selecting a notebook kernel is not the same as opening a terminal. You do not need to uninstall Anaconda or discard your STAT 201 notebooks.

## Code, Markdown, and Saved Outputs

Use code cells for computation and Markdown cells for explanations, headings, and answers. Run cells with their play controls or `Shift+Enter`. Save the notebook with `Ctrl+S` or `Cmd+S`.

A notebook remembers values from previously executed cells. Moving or editing a cell does not undo its earlier effects. Before considering work complete, **restart the kernel, run all code cells in order, and save again**. A successfully executed notebook should not depend on variables created in deleted cells.

The HTML submitted later will be generated from this saved notebook. Writing an explanation in a code comment is useful, but Markdown is better for the report's narrative.


## If Something Does Not Work

| Symptom | First check |
|---|---|
| No kernel is offered | Python/Jupyter extensions are enabled; a Python interpreter is installed and discoverable |
| Code will not run | The cell is Code rather than Markdown and a kernel is selected |
| Output differs after restarting | All required assignments/imports occur earlier in the notebook |
| You cannot find the data folder | The ZIP is extracted and the correct project folder is open |
