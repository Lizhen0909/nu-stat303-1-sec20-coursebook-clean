## Ask Three Questions Before a Command

| Question | Notebook | Terminal |
|---|---|---|
| Where am I typing? | Python code cell | Check the terminal's shell label |
| Which Python am I using? | `sys.executable` | `python -c "import sys; print(sys.executable)"` |
| Which folder is current? | `Path.cwd()` | Use the shell-specific command below |

The notebook and terminal can use **different Python environments and different working directories**. The folder visible in Explorer does not prove either process's current directory. An active-environment prompt is a hint, not proof that the notebook kernel matches.


## Navigate in the Terminal

Open **Terminal → New Terminal** in VS Code. Use the column for your shell. Windows PowerShell and Windows Command Prompt are different shells.

| Task | macOS zsh/Bash | Windows PowerShell | Windows Command Prompt |
|---|---|---|---|
| Show current directory | `pwd` | `Get-Location` | `cd` |
| List files | `ls` | `Get-ChildItem` | `dir` |
| Enter a child folder | `cd data` | `cd data` | `cd data` |
| Move to parent | `cd ..` | `cd ..` | `cd ..` |

`.` means the current directory and `..` its parent. Put paths containing spaces in quotes. To change drives as well as folders in Command Prompt, use `cd /d "D:\your\folder"`.

**Starting-folder check:** navigate to the extracted `stat303-setup` directory. Listing files should show `activity01.ipynb`, `activity02.ipynb`, and `data`. Do this before creating an environment or writing a requirements file.


## Create a Course Environment

Third-party packages extend Python: NumPy supports arrays, Pandas supports tabular data, and Matplotlib and Seaborn support visualization. Different projects can require incompatible versions, so keep their packages separate.

For these activities, use one `.venv` in `stat303-setup`. `venv` is Python's environment-creation module; `.venv` is the folder name we choose. Do not put notebooks or data inside that folder or copy it between computers.

From the VS Code Command Palette, use **Python: Create Environment** if available, choose **Venv**, and select the existing Python version supported for the course. In newer environment interfaces, use the environment manager's create/custom-create action to choose the interpreter explicitly. If `.venv` already exists for this project, inspect and reuse it rather than creating another by accident.

::: {.callout-note collapse="true" title="Alternative: create the environment in the terminal"}
**Run in the terminal, starting in `stat303-setup`.** First identify the intended Python. If `python` refers to that installation:

```text
python -c "import sys; print(sys.executable)"
python -m venv .venv
```

On macOS, your existing installation may use `python3`; substitute it in both commands before activation. If neither command identifies the intended interpreter, use the environment-creation interface or a verified full executable path. Do not guess among multiple installations.
:::

### Activate and Verify

**Starting folder: `stat303-setup`. Run only the command for your shell.**

| Shell | Activation command |
|---|---|
| macOS zsh/Bash | `source .venv/bin/activate` |
| Windows PowerShell | `.\.venv\Scripts\Activate.ps1` |
| Windows Command Prompt | `.venv\Scripts\activate.bat` |

After activation, verify in the terminal:

```text
python -c "import sys; print(sys.executable)"
python -m pip --version
```

Both should identify this project's `.venv`. If PowerShell blocks the activation script, you can use the executable directly without changing execution policy:

```powershell
.\.venv\Scripts\python.exe -m pip --version
```

On macOS, the equivalent direct executable is `.venv/bin/python`. Activation is a convenience that changes terminal command lookup; invoking the correct executable directly also works.

Now select the project's `.venv` in the **notebook kernel picker**. Selecting an interpreter for a workspace and activating a terminal do not replace checking the notebook kernel.


## Install Packages in the Intended Environment

### From the Terminal

**Starting folder: `stat303-setup`; verified `.venv` Python active.**

```text
python -m pip install ipykernel numpy pandas matplotlib
```

If using the PowerShell direct-executable route, replace `python` with `.\.venv\Scripts\python.exe` in these commands. `python -m pip` uses the pip belonging to that Python; bare `pip` can be ambiguous.

### From a Notebook

**Run in a code cell after selecting `.venv`:**

```python
%pip install seaborn
```

`%pip` targets the current IPython kernel's environment. Prefer the explicit `%` over relying on automagic. `!pip` invokes a shell command and may find a different pip. Restart the notebook kernel after installation when needed, then verify:

```python
import sys
import pandas as pd
import seaborn as sns
print(sys.executable)
print(pd.__version__)
print(sns.__version__)
```

Compare the notebook executable with the terminal's executable. A successful terminal installation does not prove that another notebook kernel can import the package. Keep installation work in a temporary setup cell or terminal, rather than rerunning installations every time an assignment is executed.


## Find Local Data with Paths

An **absolute path** begins at the filesystem root (including the appropriate drive on Windows). A **relative path** is interpreted from the process's current working directory. Forward slashes alone do not make an absolute path portable to another person's computer.

Use `pathlib` to build paths without manually joining platform-specific separators. In a **notebook code cell**:

```python
from pathlib import Path
print(Path.cwd())
data_path = Path("data") / "station_readings.csv"
print(data_path.resolve())
print(data_path.is_file())
```

For these root-level practice notebooks, the expected working directory is the extracted `stat303-setup` folder. If `is_file()` is false, inspect the displayed directory, filename, capitalization, and extracted folder layout. VS Code settings or a different notebook location can change the working directory; always check rather than assume.

Once it is true, inspect a few lines without needing data-analysis syntax:

```python
print(data_path.read_text(encoding="utf-8").splitlines()[:3])
```

A path expression describes a location; it does not establish that a file exists. `.exists()` checks for an existing path, while `.is_file()` checks specifically for a file.

### What Changes When a Notebook Moves?

If a notebook runs with `stat303-setup/notebooks` as its working directory, `Path("data")` refers to a folder inside `notebooks`. The sibling data folder would instead be:

```python
Path("..") / "data" / "station_readings.csv"
```

This works only when the actual working directory has that relationship. Do not repeatedly add `..` until a path happens to work.

The terminal's `cd` changes that terminal's directory, not an already running notebook's directory. A notebook's `%cd` changes its working directory persistently; `!cd` runs in a temporary shell and generally does not persist. For the main workflow, keep a consistent layout and explicit paths rather than scattering directory changes through the notebook.


## Record and Recreate Dependencies

**Terminal, starting in `stat303-setup`, with the project Python verified:**

```text
python -m pip freeze > requirements.txt
```

The shell writes `requirements.txt` into its current directory. `>` replaces an existing file with that name. Review the file and confirm that it records your project environment rather than Anaconda's entire base environment.

In a new compatible environment, install from it with:

```text
python -m pip install -r requirements.txt
```

Package versions help reproduce an environment but do not guarantee identical behavior across Python versions, operating systems, or external services. Record the Python version too. Incompatible requirements can cause pip's resolver to report an error; it does not simply select the last version mentioned.

Conda, Poetry, and uv provide other workflows. They are optional background here; use the single course workflow until you can verify it reliably.
