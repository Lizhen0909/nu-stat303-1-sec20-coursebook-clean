You will write assignment code and explanations in a VS Code Jupyter notebook, then use Quarto to generate the **HTML file for submission**. This chapter completes that workflow.


## Install and Verify Quarto

Install the **Quarto application** from [the official download page](https://quarto.org/docs/get-started/). Quarto is separate from Python; installing a similarly named Python package or a VS Code extension is not a substitute for the application.

Open a new VS Code terminal after installation and run:

```text
quarto --version
```

This command can run from any folder. A version number confirms that the terminal can find Quarto. If the command is not found, restart VS Code so it sees the updated PATH, then check the installer instructions before editing PATH manually. The optional Quarto VS Code extension can help with authoring, but the terminal command is the workflow used here.


## Use the Notebook Template

The [setup practice kit](downloads/setup-practice.zip) includes `activity03.ipynb` and `setup_assignment.ipynb`. Each begins with a **Raw** cell containing Quarto's YAML settings:

```yaml
---
title: "Setup Practice"
author: "Your Name"
format:
  html:
    embed-resources: true
    toc: true
    code-fold: false
---
```

Edit the title and author inside the existing Raw cell. Keep its `---` delimiters and indentation, and do not run this cell as Python. If your editor does not offer a Raw-cell control, retain the cell in the supplied template rather than recreating it as a code cell.

`embed-resources: true` embeds ordinary figures, styles, and scripts into the HTML. It does not package your Python environment or the input CSV for rerunning the analysis, and external websites or live services can still require internet access. `code-fold: false` keeps code visible for review.


## Write a Readable Notebook

Use Markdown headings to organize answers. Describe what a result means near the code and output that support it. Remove temporary installation cells and irrelevant exploratory output before submission. Keep required imports and computations in a runnable order.

The supplied plotting and data-reading code is a workflow demonstration. You will learn these libraries in later chapters; here the goal is to run it in the right environment and verify its output.


## Run, Save, and Render

Quarto normally uses an `.ipynb` file's **saved outputs**; it does not rerun its code by default. A successful render therefore does not prove the notebook ran correctly or contains current results.

Before rendering:

1. Select the intended `.venv` kernel.
2. Restart the kernel and run all code cells from top to bottom.
3. Resolve unexpected errors and check that figures appear.
4. Save the notebook.

**Terminal, starting in `stat303-setup`:** first list files and verify that `activity03.ipynb` is present. Then run:

```text
quarto render activity03.ipynb --to html
```

For this standalone practice kit, the output should be `activity03.html` beside the notebook. The kit deliberately has no `_quarto.yml` project configuration that would redirect the output elsewhere.

If a filename contains spaces, quote it:

```text
quarto render "my assignment.ipynb" --to html
```

The working directory determines how the terminal locates the notebook. It does not automatically change the directory in an existing notebook kernel. Quarto has an optional `--execute` flag for rerunning code during rendering; we use explicit notebook execution and saving as the main submission workflow.


## Inspect the HTML Before Submitting

Open the generated `.html` in a web browser. Check the title, author, explanations, code, tabular output, and figure. Look for missing outputs, stale results, raw YAML accidentally displayed as text, or unexpected error messages.

Copy **only the HTML** into a separate folder and open that copy. Ordinary plots and formatting should still appear. For this kit, disconnecting from the internet provides an additional check that the displayed report is not relying on online assets. This check concerns viewing the report, not rerunning its Python code.

Save the source notebook and input data for your own reproducible work. The integrated assignment below asks for the HTML; any additional course-platform instructions take precedence.


## Troubleshooting by Symptom

| Symptom | Check before changing anything |
|---|---|
| `ModuleNotFoundError` | Notebook `sys.executable` versus the Python used for installation |
| `FileNotFoundError` | Notebook `Path.cwd()`, resolved path, spelling, case, and extraction layout |
| No valid input files for Quarto | Terminal location and the notebook's exact filename |
| Quarto command not found | Quarto application installation and a fresh terminal/VS Code session |
| Missing or stale HTML outputs | Restart/run all, save the notebook, render again, open the latest HTML |
| HTML needs a companion folder | Raw-cell YAML includes `embed-resources: true`; render again |
