## Anaconda vs. VS Code

Anaconda and Visual Studio Code serve different roles and can be used together. Anaconda bundles Python, hundreds of packages, and environment-management tools, making it convenient but requiring more disk space. VS Code is a lightweight code editor that connects to your Python installation and lets you add extensions as needed. The table below compares their purposes and key features.

::: {.tools-comparison}

| | ![](images/anaconda-icon.svg){.comparison-logo fig-alt=""} **Anaconda** | ![](images/vscode-icon.svg){.comparison-logo fig-alt=""} **Visual Studio Code (VS Code)** |
|---|---|---|
| **Overview** | A Python distribution with environment-management tools | A powerful, flexible, and lightweight code editor |
| **What it is** | A distribution that includes Python, many common packages, and tools such as conda and Navigator | A free code editor from Microsoft, built on the open-source Code - OSS project |
| **Main purpose** | Makes it easy to install Python and manage packages and environments | Provides a modern, flexible environment to write, run, and debug code in Python and many other languages |
| **Key features** | • Preinstalls many popular data science packages<br><br>• Environment management with conda<br><br>• Anaconda Navigator graphical interface | • Clean, customizable interface<br><br>• Python support through extensions<br><br>• Integrated terminal and Jupyter notebook support through the Jupyter extension<br><br>• Debugging, code navigation, and Git tools<br><br>• Extensible through additional extensions<br><br>• Works with different Python environments, including Anaconda environments |

:::

## Check Your Existing Setup

**Already have VS Code?** Open it and verify that the **Python** and **Jupyter** extensions from Microsoft are installed and enabled. Then check for Python using the steps below.

**Need VS Code?** Install it from [the official website](https://code.visualstudio.com/), then open the Extensions panel and install Microsoft's Python and Jupyter extensions. On Windows/Linux the shortcut is `Ctrl+Shift+X`; on macOS it is `Cmd+Shift+X`.

### Find an Existing Python Installation

A **Python interpreter** is the program that runs your Python code. Since you completed STAT 201, you should already have Python installed. Here, you will connect VS Code to that existing installation.

1. Open VS Code's **Command Palette**: `Ctrl+Shift+P` on Windows/Linux or `Cmd+Shift+P` on macOS.
2. Type **Python: Select Interpreter** and select that command.

   ![Find Python in VS Code: open the Command Palette, select Python: Select Interpreter, and identify an existing installation by its version and path. Versions and paths shown are examples.](images/vscode-select-interpreter.png){#fig-select-interpreter width=100% fig-alt="Two-panel illustration of VS Code showing the Python: Select Interpreter command and a picker with an Anaconda environment. Arrows identify the command, Python version, and installation path."}

3. Look for a listed **Python 3** environment. The entry usually shows a version and an installation path; an Anaconda installation may be labeled **conda** or **base**, or have `anaconda3` or `miniconda3` in its path.
4. Select the **Python 3 environment you used in STAT 201**. If you used Anaconda's default environment, choose its **base** environment. An environment includes both a Python version and its installed packages.

- **For this chapter, choose Anaconda's `base` environment.** If you did not use Anaconda, choose the working Python 3 environment you used in STAT 201. In the next chapter, you will learn the disadvantages of using `base` for all Python projects and how to create an isolated, dedicated environment for this course.
- **Identify it by its executable path.** Open a local Jupyter notebook using your usual STAT 201 setup, run `import sys` followed by `print(sys.executable)`, and match that path to the interpreter in VS Code. Different environments can use the same Python version, so the version number alone is not enough.
- **Use the same environment for your notebook kernel.** When you open an `.ipynb` file in VS Code, use the kernel picker at the top right to choose that environment as well. Selecting an interpreter and selecting a notebook kernel are separate steps.
- **If you still cannot identify it, ask for setup assistance.** Share the listed environment names and paths; you do not need to install another copy of Python just because several appear.

### Check That It Runs

Finding an interpreter in the list is a first check. Next, verify that it can execute code:

1. Open the Command Palette again and run **Python: Start Terminal REPL**. This starts an interactive Python session using the selected interpreter.
2. When the `>>>` prompt appears, enter these lines **one at a time**, pressing Enter after each. Do not type the `>>>` prompt itself.

```python
import sys
print(sys.version)
print(sys.executable)
print(2 + 3)
```

You should see a **Python 3 version**, the **path to its executable**, and **`5`**. This confirms that the selected Python installation runs. Keep the executable path so you can identify the same environment later. Type `exit()` and press Enter to leave Python, then continue to **Open a Project Folder** below.

### If No Interpreter Appears or Python Does Not Start

An empty list does **not** necessarily mean Python is missing; VS Code may not have found it.

- **No “Python: Select Interpreter” command?** Check that Microsoft's Python extension is installed and enabled, then restart VS Code.
- **Previously used Anaconda?** Open Anaconda Navigator and try launching a local Jupyter notebook as you did in STAT 201. Run `import sys` followed by `print(sys.executable)` in a code cell. If it works, Python is installed. Copy that path, return to **Python: Select Interpreter**, and use **Enter interpreter path…** if available to locate that executable.
- **Still cannot find or run Python?** Ask for setup assistance and share the error message. If no working installation is available, install the instructor's supported Python version from the [official Python website](https://www.python.org/downloads/).

## Code, Markdown, and Saved Outputs

Use code cells for computation and Markdown cells for explanations, headings, and answers. Run cells with their play controls or `Shift+Enter`. Save the notebook with `Ctrl+S` or `Cmd+S`.

A notebook remembers values from previously executed cells. Moving or editing a cell does not undo its earlier effects. Before considering work complete, **restart the kernel, run all code cells in order, and save again**. A successfully executed notebook should not depend on variables created in deleted cells.

The HTML submitted later will be generated from this saved notebook. Writing an explanation in a code comment is useful, but Markdown is better for the report's narrative.
